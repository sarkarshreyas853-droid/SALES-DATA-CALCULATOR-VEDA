# Task 4: Sales Data Calculator

This notebook analyses a manually created collection of sales values. It calculates the total sales, average sale, highest sale and lowest sale, and checks the results against the original data.

```
# Manually created sales data (amount in rupees, one entry per sale)
sales = [12500, 8400, 15300, 9800, 22100, 7600, 13450, 18900, 11200, 16750]
print("Original data:", sales)
print("Number of sales:", len(sales))
```

## Sales statistics

The calculations use built-in Python functions: `sum`, `max` and `min`. The average is the total divided by the number of sales.

```
total_sales = sum(sales)
average_sale = total_sales / len(sales)
highest_sale = max(sales)
lowest_sale = min(sales)

print(f"Total sales:   {total_sales:,}")
print(f"Average sale:  {average_sale:,.2f}")
print(f"Highest sale:  {highest_sale:,}")
print(f"Lowest sale:   {lowest_sale:,}")
```

## Check against the original data

Each result is compared with the original list, so the calculations are verified rather than assumed.

```
assert total_sales == 12500 + 8400 + 15300 + 9800 + 22100 + 7600 + 13450 + 18900 + 11200 + 16750
assert highest_sale in sales and lowest_sale in sales
assert lowest_sale <= average_sale <= highest_sale
assert abs(average_sale * len(sales) - total_sales) < 1e-6
print("All checks passed: the total, average, highest and lowest sale agree with the original data.")
```

## Explanation of the results

The total of all sales is the sum of every value. The average is that total spread across the number of sales, so it describes a typical sale. The highest and lowest values show the range, so you can see how widely sales vary.

## Answers to the interview questions

1. **How can Python help analyse sales data?** Python can store the values, calculate totals and averages with built-in functions, and scale to large datasets with libraries such as Pandas.
2. **What does average sales represent?** The average is the typical value of one sale: the total sales divided by the number of sales.
3. **How would you find the highest sale?** Use `max(sales)`. With Pandas, use `df['sale'].max()`.
